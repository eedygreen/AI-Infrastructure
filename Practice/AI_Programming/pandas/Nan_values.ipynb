```python
import pandas as pd
```

```python
items = [{'bikes': 20, 'pants': 30, 'watches': 35, 'shirts': 15, 'shoes':8, 'suits':45},
          {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants':5, 'shirts': 2, 'shoes':5, 'suits':7},
          {'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4, 'shoes':10}]
```

```python
store_items = pd.DataFrame(items, index=['store 1', 'store 2', 'store 3'])
store_items
```

```python
store_items.isnull()
```

```python
find_null_items = store_items.isnull().sum().sum()
print(find_null_items)
```

```python
find_null_items = store_items.isnull().sum()
print(find_null_items)
```

```python
print('Number of non-NaN values in the columns of our DataFrame:\n', store_items.count())
```

#### Removing Nan Values

```python
store_items.dropna(axis=0)
```

```python
store_items.dropna(axis=1)
```

#### Substituting Nan Values

```python
store_items.fillna(0)
```

#### Forward filling

```python
store_items.ffill(axis=0)
```

Forward Filling for Columns

```python
store_items.ffill(axis=1)
```

Backward Filling

```python
store_items.bfill(axis=0)
```

Backward Filling for Columns

```python
store_items.bfill(axis=1)
```

Noticed Glasses is not filled, that is because there is no item after it that be backward filled

Interpolate 

```python
store_items.interpolate(axis=0)
```

```python
store_items.interpolate(axis=1)
```
