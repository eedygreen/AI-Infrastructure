```python
import pandas as pd
```

```python
books = pd.Series(data = [1, 1, 1, 1, 18, 15, 4], index = ["Mind for Numbers", "Distributed Service with go", "Mathematical thinking", "A Staff Engineer's Path", "Nur Aalam Numbalai", "lisaanu Al'arab", "Sahih Ibn Kathir"] )
```

```python
books
```

```python
print('shape', books.shape)
print('Dinmension', books.ndim)
print('size', books.size)
```

```python
books.index
```

```python
books.values
```

### Check for an Index or Value

```python
check_assuyuti = 'assuyuti' in books
print(check_assuyuti)
```

```python
print(books['Mind for Numbers'])
print(books[['Mind for Numbers', 'Distributed Service with go']])
print(books.loc[['Mind for Numbers', 'Distributed Service with go']])
print(books[[0,1]]) #numerical access
print(books[[-1]])
print(books[0])
print(books.iloc[[2,3]])
```
