```python
items2 = [{'bikes': 20, 'pants': 30, 'watches': 35}, {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants':5}]
```

```python
store = pd.DataFrame(items2, index=['store1', 'store2'])
```

```python
store
```

```python
print('How many bikes are in each store\n', store['bikes'])
```

```python
print('How many bikes are in store1\n', store['bikes']['store1'])
```

```python
print('How many watches and bikes', store[['watches', 'bikes']])
```

```python
print('How many watches and bikes in store1', store.loc['store1'])
```

#### Create a New row and add it to store

```python
new_items = [{'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4}]
```

```python
new_store = pd.DataFrame(new_items, index=['store3'])
new_store
```

```python
store = pd.concat([store, new_store])
store
```
