```python
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
fuel_econ = pd.read_csv('./data/fuel_econ.csv')
fuel_econ.head()
```

```python
fuel_econ['displ'].describe()
```

```python
plt.figure(figsize=[20,5])
plt.subplot(1,2,1)
sb.regplot(data = fuel_econ, x = 'displ', y = 'comb', truncate = False, x_jitter = 0.04, scatter_kws = {'alpha': 1/10}, fit_reg = False);
plt.xlabel('Combined Fuel Efficiencies in mpg')
plt.ylabel('Displacement')

plt.subplot(1,2,2)
plt.hist2d(data = fuel_econ, x = 'displ', y = 'comb')
plt.colorbar()
plt.xlabel('Combined Fuel Efficiencies in mpg')
plt.ylabel('Displacement')

plt.show()
```

```python

plt.hist2d(data = fuel_econ, x = 'displ', y = 'comb', cmap = 'viridis_r', cmin=0.5)
plt.colorbar()
plt.xlabel('Combined Fuel Efficiencies in mpg')
plt.ylabel('Displacement')

plt.show()
```

Add Text Annotation to each cell

```python
bin_x = np.arange(0.6, 7 + 0.7, 0.7)
bin_y = np.arange(12, 58 + 10, 10)

h2d = plt.hist2d(data = fuel_econ, x = 'displ', y = 'comb', cmap = 'viridis_r', cmin=0.5, bins = [bin_x, bin_y])
plt.colorbar()
plt.xlabel('Combined Fuel Efficiencies in mpg')
plt.ylabel('Displacement')

counts = h2d[0]
for i in range(counts.shape[0]):
    for j in range(counts.shape[1]):
        c = counts[i, j]
        if c >= 200:
            plt.text(bin_x[i] + 0.5, bin_y[i] + 0.5, int(c), ha = 'center', va = 'center', color = 'white')
        elif c > 0:
            plt.text(bin_x[i] + 0.5, bin_y[i] + 0.5, int(c), ha = 'center', va = 'center', color = 'black')

plt.show()
```
