```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
%matplotlib inline
```

```python
pkm = pd.read_csv('./data/pokemon.csv')
```

```python
pkm.head()
```

```python
pkm['height'].describe()
```

```python
bins = np.arange(0, pkm['height'].max() + 0.1, 0.1)
```
