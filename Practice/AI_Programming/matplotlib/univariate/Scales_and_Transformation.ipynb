```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
%matplotlib inline
```

```python
pkm = pd.read_csv('./data/pokemon.csv')
pkm.head()
```

```python
pkm['weight'].describe()
```

# Log normal Distribution

# clearly this wont fit into normal distribution
But logarithmic distribution 

```python
plt.figure(figsize=[20,5])
```

```python
plt.subplot(1,2,1)
plt.hist(data = pkm, x='weight');
plt.xlabel('Initital Plot')

plt.subplot(1,2,2)
bins = np.arange(0, pkm['weight'].max() + 40, 40) 
plt.hist(data=pkm, x='weight', bins = bins);
plt.xscale('log');
plt.xlabel('The X axis scaled to log type')
```

```python
plt.show()
```

```python
pkm['weight'].describe()
```

```python
np.log10(pkm['weight'].describe())
```

```python
# Axis Transformation
bins = 10 ** np.arange(-1, 3 + 0.1, 0.1) 
plt.hist(data=pkm, x='weight', bins = bins);
plt.xscale('log')
plt.xlabel('X Axis limits scaled and changed to log type')
```

```python
plt.show()
```

The log-normal distribution is better than the initial and scaled log-type
but the x-axis is difficult to interprete

```python
bins = 10 ** np.arange(-1, 3 + 0.1, 0.1) 
ticks = [0.1, 0.3, 1, 3, 10, 30, 100, 300, 1000]
labels = ['{}'.format(v) for v in ticks]
plt.hist(data=pkm, x='weight', bins = bins);
## The argument in the xscale() represents the axis scale type to apply.
## The possible values are: {"linear", "log", "symlog", "logit", ...}
plt.xscale('log')
plt.xticks(ticks, labels);
plt.show()
```
