```python
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
pkmn = pd.read_csv('./data/pokemon.csv')

pkmn.head()
```

```python
plt.hist(data=pkmn, x='speed');
plt.show()
```

Guage the minimum and maximum bin that will be approprait for this histogram

```python
pkmn['speed'].describe()
```

Minimum is 5.0 and max is 160.0

```python
plt.hist(data=pkmn, x='speed', bins = 20 );
plt.show()
```

Calculate the bins ranges

```python
bins = np.arange(0, pkmn['speed'].max() + 5, 5)
```

```python
plt.hist(data=pkmn, x='speed', bins = bins );
plt.show()
```

Alternative is to use the seaborns' Histogram

```python
sb.histplot(data = pkmn, x = 'speed', bins = bins, stat = 'density', kde = True);
plt.show()
```

Two graph side by side

```python
plt.figure(figsize=[20, 5])

# Histogram on the left: example of too-large bin size
# 1 row, 2 columns, subplot 1
plt.subplot(1, 2, 1)
bins = np.arange(0, pkmn['speed'].max() + 4, 4)
plt.hist(data=pkmn, x='speed', bins=bins)

# Histogram on the right: example of too-small bin size
# 1 row, 2 columns, subplot 2
plt.subplot(1, 2, 2)
bins = np.arange(0, pkmn['speed'].max() + 0.25, 0.25)
plt.hist(data=pkmn, x='speed', bins=bins)

plt.show()
```

```python
fig, axes = plt.subplots(3, 4) # grid of 3x4 subplots
axes = axes.flatten() # reshape from 3x4 array into 12-element vector
for i in range(12):
    plt.sca(axes[i]) # set the current Axes
    plt.text(0.5, 0.5, i+1) # print conventional subplot index number to middle of Axes
```
