#### Slicing 2D Array Operations

```python
o = np.arange(20).reshape(4,5)
print(o)
```

```python
selected_rows = o[1:4,2:5]
print(selected_rows)
```

```python
same_as_selected_row = o[1:,2:5]
print(same_as_selected_row)
```

```python
third_row_2_4_column = o[:3,2:5]
print(third_row_2_4_column)
print('Third row', o[2,:])
print('Third column', o[:,2])
print('Third column with rank2 array', o[:,2:3])
```

### Array modification
An array created from an array is a copy/view of the original array, which means any modification will reflect on either copy

```python
st = np.arange(30).reshape(5,6)
print(st)
```

```python
sq = st[1:3,3:5]
print(sq)
```

```python
sq[1,0] = 512

print('modifying a copy modifies st', st)
print('modifying sq modifies the st', sq)
```

#### Copy - with copy the underlying array is not modified

```python
sq_copy = np.copy(st[2:,3:])
print(sq_copy)
print(st)
```

```python
sq_copy[2,0] = 600
print(sq_copy)
print(st)
```

#### Indices - using array to create indices

```python
indices = np.array([1,2])
print(indices)
```

```python
slice_with_indices = st[indices,:]
same_as = st[1:3,:]
print(slice_with_indices)
print(same_as)
```

#### Use an array as indices to extract specific rows from a rank 2 ndarray

```python
extract_with_indices = np.random.randint(1,20, size=(50,5))
print(extract_with_indices)
print(extract_with_indices.shape)
```

```python
# Create a rank 1 ndarray that contains a randomly chosen 10 values between `0` to `len(X)` (50)
# The row_indices would represent the indices of rows of X
row_indices = np.random.randint(0,50, size=10)
print("Random 10 indices are: ", row_indices)
```

```python
print(extract_with_indices[row_indices, :])
print()
print(extract_with_indices[row_indices[4:5], :])
```
