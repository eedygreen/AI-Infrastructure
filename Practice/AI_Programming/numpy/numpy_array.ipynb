```
linspace is a Better alternative to arange in the case of floating types
usage
numpy.linspace(start, stop, num=50, endpoint=True, retstep=False, dtype=None, axis=0)
```

````python
v = np.linspace(0,25,10)
print(v)
````

`exluding the stop point use the endpoint = False`

````python
vs = np.linspace(0,25,10, endpoint = False)
print(vs)
````

The np.reshape(ndarray, new_shape) function converts the given ndarray into the specified new_shape. It is important to note that the new_shape should be compatible with the number of elements in the given ndarray.

````python
new_shape = np.reshape(vs, (2,5))
print(new_shape)
````

reshape() is a method, thus it can be called directly

````python
wv = np.linspace(0,25,10, endpoint=False).reshape(2,5)
print(wv)
````
