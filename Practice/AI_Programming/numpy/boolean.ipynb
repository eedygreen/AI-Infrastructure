```python
import numpy as np
```

### Boolean slicing with Array

```python
n = np.arange(25).reshape(5,5)
print(n)
```

```python
print('Elements of n greater than 10: ', n[(n>10)])

print('\nElements of n greater than or equal to 15: ', n[(n>=15)])

print('\nElements of n greater than 12 and less than  15: ', n[(n > 12) & (n < 15)])
```

##### Use Boolean Indexing to assign value

```python
n[(n > 13) & (n < 20)] = -1
print(n)
```

### Set Operations

```python
p = np.array([2,5,6,8,10])

q = np.array([4,6,8,10,12])
```

```python
print('Elements that are both p and q', np.intersect1d(p,q))
print('\nElements that are not in p and q', np.setdiff1d(p,q))
print('\nAll Elements of both p and q', np.union1d(p,q)) 
```
