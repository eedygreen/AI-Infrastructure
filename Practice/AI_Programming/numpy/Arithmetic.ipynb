```python
import numpy as np
```

```python
x = np.array([1,2,3,4])
y = np.array([5.5,6.5,7.5,8.5])
```

```python
print('add x, y', x + y)
print('add x, y with numpy numpy function', np.add(x,y))
```

```python
print('subtract x, y', x - y)
print('subtract x, y with numpy numpy function', np.subtract(x,y))
```

#### operation on 2D array

```python
m = np.array([1,2,3,4]).reshape(2,2)
n = np.array([5.5,6.5,7.5,8.5]).reshape(2,2)
```

```python
print('add x, y', m + n)
print('add x, y with numpy numpy function', np.add(m,n))

print('\nSubtract x, y', m - n)
print('Subtract x, y with numpy numpy function', np.subtract(m,n))

print('\nMulitply x, y', m * n)
print('Multiply x, y with numpy numpy function', np.multiply(m,n))

print('\nDivide x, y', m / n)
print('Divide x, y with numpy numpy function', np.divide(m,n))
```

```python
print(m)
print('sqaure root', np.sqrt(m))
print('square', np.power(m, 2))
print('exponential', np.exp(m))

print('\nAverage', np.mean(m))
print('Average of all rows', np.mean(m, axis=1))
print('Average of all columns', np.mean(m, axis=0))
```

```python
print(m)
print('\nMedian', np.median(m))
print('Median of all rows', np.median(m, axis=1))
print('Median of all columns', np.median(m, axis=0))

print('\nStandard Deviation', np.std(m))
print('Standard Deviation of all rows', np.std(m, axis=1))
print('Standard Deviation of all columns', np.std(m, axis=0))

print('\nSum', np.sum(m))
print('Sum of all rows', np.sum(m, axis=1))
print('Sum of all columns', np.sum(m, axis=0))

print('\nmin', np.min(m))
print('min of all rows', np.min(m, axis=1))
print('min of all columns', np.min(m, axis=0))

print('\nmax', np.max(m))
print('max of all rows', np.max(m, axis=1))
print('max of all columns', np.max(m, axis=0))
```

### Statistical Functions

```python
print(m)
print('\nMedian', np.median(m))
print('Median of all rows', np.median(m, axis=1))
print('Median of all columns', np.median(m, axis=0))

print('\nStandard Deviation', m.std())
print('Standard Deviation of all rows', m.std(axis=1))
print('Standard Deviation of all columns', m.std(axis=0))

print('\nSum', m.sum())
print('Sum of all rows', m.sum(axis=1))
print('Sum of all columns', m.sum(axis=0))

print('\nmin', np.min(m))
print('min of all rows', m.min(axis=1))
print('min of all columns', m.min(axis=0))

print('\nmax', np.max(m))
print('max of all rows', m.max(axis=1))
print('max of all columns', m.max(axis=0))
```
